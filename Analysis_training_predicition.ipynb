Using Logistic Regression model for binary classification

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, r2_score, mean_absolute_error,mean_squared_error,accuracy_score
from sklearn.model_selection import train_test_split
```

```python
x= df[['Pclass','Age','SibSp', 'Parch','Fare','Sex_male','Embarked_Q','Embarked_S','GroupSize','IndividualFare']]
y=df['Survived']
```

```python
x_train, x_test, y_train,  y_test = train_test_split(x,y,test_size=0.2,random_state=40)
```

```python
LogReg = LogisticRegression(max_iter=1000)
LogReg.fit(x_train,y_train)
```

```python
y_pred = LogReg.predict(x_test)
```

```python
print(f'Confusion matrix \n{confusion_matrix(y_test,y_pred)}')
print(f'Classification report \n {classification_report(y_test,y_pred)}')
print(f"Accuracy: {accuracy_score(y_test, y_pred):.2f}")
```
